```python
!apt-get -qq install hmmer
import glob
PFAM_DIR = glob.glob(f"{DATA_DIR}/raw/shared/interpro_pfam/*")[0]
hmm_gz = f"{PFAM_DIR}/Pfam-A.hmm.gz"
hmm_path = hmm_gz[:-3]
if not os.path.exists(hmm_path):
    !gunzip -k "{hmm_gz}"
if not os.path.exists(f"{hmm_path}.h3m"):
    !hmmpress -f "{hmm_path}"
```

```python
cpu = os.cpu_count()
protein_fasta = glob.glob(f"{DATA_DIR}/raw/oryza_sativa/ncbi_refseq_oryza_irgsp/*/*_protein.faa.gz")[0]

cmd = (f'python scan_pfam_domains.py --pfam-hmm "{hmm_path}" --proteins "{protein_fasta}" '
       f'--species oryza_sativa --assembly "IRGSP-1.0" '
       f'--out-dir "{DATA_DIR}/pfam_scan/oryza_sativa" --cpu {cpu} --chunk-size 2000')
print("$", cmd)
!{cmd}
```
